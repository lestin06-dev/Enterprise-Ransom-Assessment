# Enterprise Ransomware Risk Assessment Tool

**Purpose:** A defensive, beginner-friendly tool that scores hypothetical enterprise ransomware risks and prioritizes mitigation actions.

**Scope and safety:** This project does not create, run, or analyze ransomware. It uses an illustrative risk register for educational purposes only. Scores are subjective examples, not measured probabilities or an audit of a real organization.


## 1. Risk scoring method

- Likelihood: 1 (rare) to 5 (almost certain)
- Impact: 1 (minimal) to 5 (severe)
- Risk score = Likelihood × Impact
- 1–4 = Low; 5–9 = Moderate; 10–16 = High; 17–25 = Critical

These thresholds are a simple demonstration model. An organization should tailor them to its risk appetite and context.

```python
import csv
from pathlib import Path

risks = [
    {'id': 'R01', 'asset': 'Employee endpoints', 'threat': 'Phishing email leads to credential theft or malware execution', 'likelihood': 4, 'impact': 5, 'existing_controls': 'Email filtering, endpoint protection, phishing awareness training, MFA', 'recommendation': 'Run regular awareness exercises; verify endpoint protection coverage; enforce MFA.'},
    {'id': 'R02', 'asset': 'Shared business files', 'threat': 'Ransomware encrypts shared files and disrupts operations', 'likelihood': 4, 'impact': 5, 'existing_controls': 'Access controls, segmented shares, offline/immutable backups', 'recommendation': 'Test restore procedures quarterly; restrict write access; maintain protected backups.'},
    {'id': 'R03', 'asset': 'User accounts', 'threat': 'Stolen passwords are used to access business systems', 'likelihood': 3, 'impact': 5, 'existing_controls': 'Multi-factor authentication, password manager, least privilege', 'recommendation': 'Enable MFA for remote and privileged access; review dormant accounts monthly.'},
    {'id': 'R04', 'asset': 'Backup repository', 'threat': 'Attacker deletes or encrypts accessible backups', 'likelihood': 3, 'impact': 5, 'existing_controls': 'Offline or immutable backup copies, separate admin accounts', 'recommendation': 'Keep at least one isolated backup copy and conduct documented recovery tests.'},
    {'id': 'R05', 'asset': 'Business applications', 'threat': 'Unpatched software is exploited to gain a foothold', 'likelihood': 3, 'impact': 4, 'existing_controls': 'Asset inventory, patch management, vulnerability remediation', 'recommendation': 'Set patch timelines by severity and track overdue critical updates.'},
    {'id': 'R06', 'asset': 'Network and servers', 'threat': 'Malware spreads laterally across systems', 'likelihood': 3, 'impact': 5, 'existing_controls': 'Network segmentation, EDR, restricted admin privileges', 'recommendation': 'Separate critical systems; monitor unusual authentication and file activity.'},
    {'id': 'R07', 'asset': 'Incident response process', 'threat': 'Delayed detection increases downtime and recovery cost', 'likelihood': 3, 'impact': 4, 'existing_controls': 'Centralized logging, alerting, response playbooks', 'recommendation': 'Assign incident roles and rehearse a ransomware tabletop exercise.'},
    {'id': 'R08', 'asset': 'Third-party services', 'threat': 'Compromised vendor account or service affects operations', 'likelihood': 2, 'impact': 4, 'existing_controls': 'Vendor review, scoped access, MFA, contractual security requirements', 'recommendation': 'Review critical suppliers and remove third-party access when no longer needed.'},
    {'id': 'R09', 'asset': 'Sensitive business data', 'threat': 'Data exfiltration causes privacy, legal, or reputational harm', 'likelihood': 3, 'impact': 5, 'existing_controls': 'Data classification, access logging, DLP where appropriate', 'recommendation': 'Limit data access, monitor bulk downloads, and maintain a breach notification plan.'},
    {'id': 'R10', 'asset': 'Staff awareness', 'threat': 'Employees report suspicious activity too late', 'likelihood': 3, 'impact': 3, 'existing_controls': 'Reporting channel, awareness training, clear escalation route', 'recommendation': 'Provide a simple reporting process and reinforce prompt reporting without blame.'},
]

def classify_risk(score):
    if score >= 17:
        return 'Critical'
    if score >= 10:
        return 'High'
    if score >= 5:
        return 'Moderate'
    return 'Low'

for risk in risks:
    risk['score'] = risk['likelihood'] * risk['impact']
    risk['risk_level'] = classify_risk(risk['score'])

risks_sorted = sorted(risks, key=lambda item: item['score'], reverse=True)
print(f'Number of illustrative risks assessed: {len(risks_sorted)}')
print('Highest score:', risks_sorted[0]['score'], '| Level:', risks_sorted[0]['risk_level'])
```

## 2. View the prioritized risk register
The following cell prints a compact register using only Python's standard library.

```python
columns = ['id', 'asset', 'likelihood', 'impact', 'score', 'risk_level']
widths = {col: max(len(col), max(len(str(row[col])) for row in risks_sorted)) for col in columns}
print(' | '.join(col.upper().ljust(widths[col]) for col in columns))
print('-+-'.join('-' * widths[col] for col in columns))
for row in risks_sorted:
    print(' | '.join(str(row[col]).ljust(widths[col]) for col in columns))
```

## 3. Export results to CSV
Run this cell to create `ransomware_risk_register.csv` in the notebook's working directory.

```python
output_file = Path('ransomware_risk_register.csv')
fieldnames = ['id', 'asset', 'threat', 'likelihood', 'impact', 'score', 'risk_level', 'existing_controls', 'recommendation']
with output_file.open('w', newline='', encoding='utf-8') as csvfile:
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    for risk in risks_sorted:
        writer.writerow({field: risk[field] for field in fieldnames})
print(f'CSV exported to: {output_file.resolve()}')
```

## 4. Interpretation and action priorities

1. Treat the highest-scoring risks first, especially threats affecting shared files, endpoints, identity, and backups.
2. Validate that protected backups exist and test restoring them; a backup that has never been tested may not support recovery.
3. Reduce the chance of account compromise with multi-factor authentication and least-privilege access.
4. Maintain patching, endpoint protection, logging, incident-response roles, and staff reporting procedures.

The scores support discussion; they do not replace expert judgment or a formal risk assessment.

## 5. Limitations and responsible use

- The register is hypothetical and contains no real company data.
- Likelihood and impact ratings are illustrative judgments, not statistical estimates.
- The tool does not inspect systems, scan networks, detect malware, or guarantee security.
- Before real-world use, have qualified staff validate assets, threats, controls, scoring criteria, and legal/privacy requirements.
